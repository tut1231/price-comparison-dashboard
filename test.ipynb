```python
import requests
from bs4 import BeautifulSoup
url = 'https://24h.pchome.com.tw/prod/DPAD06-A900KOI8D'
response = requests.get(url)
print(response.status_code)
print(response.encoding)
```

```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(response.text, 'lxml')
print(soup.prettify())
```

```python
main_titles = soup.find_all('div', 'c-blockCombine__title')
for title in main_titles:
    text = title.get_text().strip()
    if text != "規格說明":
        continue
    print(text)
articles_obj = soup.find_all('tr','c-tableGrid__row c-tableGrid__row--borderBottom')

for obj in articles_obj:
    th = obj.find('th','c-tableGrid__th c-tableGrid__th--transparentLeft')
    td = obj.find('td','c-tableGrid__column')
    if th and td:
       title = th.get_text().strip()
       value = td.get_text().strip()
       print(f"{title} : {value}")
c_blockCombine = soup.find_all('div','c-blockCombine c-blockCombine--gray')
for blockCombines in c_blockCombine:
    sub_Title = blockCombines.find('div','c-blockCombine__subTitle')
    item = blockCombines.find('div','c-blockCombine__item')
    htmlText = blockCombines.find('div','c-blockCombine__htmlText')
    continue_tag = item or htmlText
    if sub_Title and continue_tag:
        Title = sub_Title.get_text().strip()
        Value = continue_tag.get_text().strip()
        print(f'{Title} :')
        print(Value)
```
